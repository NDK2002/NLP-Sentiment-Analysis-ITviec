# 🧹 Giai đoạn 2: Tiền xử lý & Chuẩn hóa văn bản Tiếng Việt
Áp dụng bộ từ điển teencode, emoji, lỗi chính tả và tách từ underthesea


```python
import sys, os
sys.path.append("..")
import pandas as pd
from src.preprocessing import TextPreprocessor
from tqdm import tqdm
tqdm.pandas()

# Khởi tạo bộ tiền xử lý
preprocessor = TextPreprocessor()

# Đọc dữ liệu
df = pd.read_excel("../data/raw/Reviews.xlsx")
print(f"Kích thước dữ liệu gốc: {df.shape}")
```

## 2.1 Ghép các trường văn bản và làm sạch


```python
# Ghép các cột review nếu cần (vd: Title + What I liked + Suggestions)
df["clean_text"] = df["What I liked"].fillna("") + " " + df["Suggestions for improvement"].fillna("")
df["processed_text"] = df["clean_text"].progress_apply(lambda x: preprocessor.tokenize_vietnamese(x, remove_stopwords=True))

# Lưu kết quả ra processed/
os.makedirs("../data/processed", exist_ok=True)
df.to_excel("../data/processed/reviews_cleaned.xlsx", index=False)
print("Đã hoàn thành tiền xử lý và lưu tại data/processed/reviews_cleaned.xlsx")
```
