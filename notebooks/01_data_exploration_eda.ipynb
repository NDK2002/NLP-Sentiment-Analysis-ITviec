# 📊 Giai đoạn 1: Khám phá & Phân tích dữ liệu (EDA)
Đồ án Phân tích Cảm xúc đánh giá ITviec


```python
import os, sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc dữ liệu thô
raw_path = "../data/raw/Reviews.xlsx"
df = pd.read_excel(raw_path)
print(f"Tổng số mẫu đánh giá: {len(df)}")
df.head()
```

## 1.1 Kiểm tra thông tin các trường & dữ liệu khuyết thiếu


```python
df.info()
print("\nSố lượng giá trị khuyết thiếu (NaN):")
print(df.isnull().sum())
```

## 1.2 Phân bố điểm đánh giá (Rating Stars)


```python
plt.figure(figsize=(8, 4))
col_target = "Rating" if "Rating" in df.columns else df.columns[0]
sns.countplot(data=df, x=col_target, palette="viridis")
plt.title("Phân bố điểm đánh giá sao")
plt.show()
```
